# SDSS Spectra: Acquisition & Dual-Format Preprocessing
### Companion to `reproduce_figure3_ppplot.ipynb` — Jespersen et al. (2025)
### *"The optical and infrared are connected"* ([arXiv:2503.03816](https://arxiv.org/abs/2503.03816))

---

## What this notebook does

This notebook downloads SDSS galaxy spectra and saves them in **two complementary
formats**, because different downstream tasks require different preprocessing:

| Output file | Format | Used by |
|---|---|---|
| `spectra_obs_frame.npy` | Observed-frame, on spender's log-λ grid | **spender encoder** |
| `spectra_restframe_unnorm.npy` | Rest-frame, NOT normalised | Manual norm_const, inspection |
| `spectra_restframe_norm.npy` / `spectra_matrix.npy` | Rest-frame, normalised | PCA, k-means clustering |
| `norm_consts.npy` | Scalar per galaxy: median flux 5300–5850 Å | MLP input feature |
| `wave_obs_grid.npy` | spender SDSS wavelength axis | Axis labels |
| `wave_grid.npy` | Rest-frame wavelength axis | Axis labels |
| `metadata.npy` | specobjid, z, petroMag_r … | Matching, alignment |

---

## Why two formats? A note for the mathematically inclined

**spender** (Melchior et al. 2023; Liang et al. 2023) is a convolutional autoencoder
trained on SDSS spectra *as observed*: it receives each galaxy's flux
on the fixed SDSS spectrograph wavelength grid
(\(10^{3.578}\)–\(10^{3.970}\) Å, 3921 log-spaced pixels)
and internally accounts for redshift in its decoder.

The **MLP** (§3.1 of the paper) does NOT receive raw spectra.  It receives:

1. The 6-dimensional spender latent vector \(\mathbf{s}\),
2. A *normalisation constant* \(c\) = the median flux in rest-frame 5300–5850 Å, and
3. The spectroscopic redshift \(z\).

So \(c\) and \(\mathbf{s}\) are computed **separately**:
- \(c\) comes from the rest-frame unnormalised spectrum.
- \(\mathbf{s}\) comes from spender applied to the observed-frame spectrum.

The MLP output (predicted WISE magnitudes) is then **multiplied by** \(c\) after
the final linear layer, so the network learns galaxy *shape* while \(c\) carries the
overall brightness.

---

**Two operating modes** (same as `wise_data_preprocessing.ipynb`):

| Mode | Trigger | Best for |
|---|---|---|
| `astroquery` | `DATA_SOURCE = "astroquery"` | Prototyping (≤ 10k objects) |
| CSV | `DATA_SOURCE = "csv"` | Full scale (510k, from CasJobs) |

## Cell 1 — Configuration

```python
# ═══════════════════════════════════════════════════════════════════════
# SCALE KNOBS — change N_SPECTRA to control run size
# ═══════════════════════════════════════════════════════════════════════
N_SPECTRA   = 200         # None → ALL (~510k). Start small.
DATA_SOURCE = "astroquery"  # "astroquery" (live query) or "csv" (CasJobs file)
CSV_PATH    = Path("sdss_output/sdss_metadata_casjobs.csv")
# ═══════════════════════════════════════════════════════════════════════

Z_MAX = 0.5   # Maximum redshift (paper uses z < 0.5)

# ── Rest-frame wavelength grid (for normalised + unnormalised versions) ─
# This is the standard linear rest-frame grid used in our clustering pipeline.
# Range 3800–9200 Å covers the key optical features at the typical SDSS z range.
WAVE_REST_MIN   = 3800.0
WAVE_REST_MAX   = 9200.0
N_PIXELS_REST   = 3600     # 1.5 Å / pixel — intentionally matches sdss_spectra_clustering.ipynb
wave_rest_grid  = np.linspace(WAVE_REST_MIN, WAVE_REST_MAX, N_PIXELS_REST)

# ── spender SDSS observed-frame wavelength grid ─────────────────────────
# spender's SDSS Instrument class defines its grid as:
#     10 ** torch.arange(3.578, 3.97, 0.0001)
# This gives 3921 log-spaced pixels from ~3784 Å to ~9333 Å in the OBSERVED frame.
# We reproduce this here so that we can save spectra on exactly this grid,
# ready to feed straight into model.encode() without further interpolation.
WAVE_OBS_LOG_START = 3.578
WAVE_OBS_LOG_STEP  = 0.0001
WAVE_OBS_LOG_END   = 3.970
wave_obs_loglam   = np.arange(
    WAVE_OBS_LOG_START,
    WAVE_OBS_LOG_END + WAVE_OBS_LOG_STEP / 2,  # +half step avoids floating-point fence-post
    WAVE_OBS_LOG_STEP,
)
wave_obs_grid = 10 ** wave_obs_loglam  # shape: (3921,)
N_PIXELS_OBS  = len(wave_obs_grid)

# ── Normalisation window (paper §3.1) ────────────────────────────────
# The MLP uses the median flux in this REST-FRAME window as the 7th input feature.
# The window is chosen because it sits on a relatively featureless continuum
# (between the Mg I b triplet ~5175 Å and the Na I D doublet ~5893 Å),
# making it a stable proxy for overall brightness.
NORM_WAVE_MIN = 5300.0
NORM_WAVE_MAX = 5850.0

# ── Sky emission lines (OBSERVED frame) ──────────────────────────────
# These are bright atmospheric emission lines that contaminate the spectra.
# They are masked *before* preprocessing but note: for the observed-frame
# spectra fed to spender, we only mask the worst pixel-quality flags
# (ivar ≤ 0, andmask ≠ 0) — spender is trained to be robust to skylines.
# For the REST-FRAME normalised spectra (for PCA), we additionally mask these.
SKYLINES_OBS    = np.array([
    5577.3, 5889.9, 6157.0, 6300.3, 6363.8,
    6834.3, 6864.0, 7340.9, 7523.7, 7715.0,
    8344.6, 8399.0, 8430.0, 8791.2, 8827.1, 9375.0,
])
SKYLINE_WIDTH_AA = 10.0   # half-width in Å — pixels within ±10 Å are masked

OUT_DIR = Path("sdss_output")
OUT_DIR.mkdir(exist_ok=True)

log.info(
    f"N_SPECTRA={N_SPECTRA}, DATA_SOURCE='{DATA_SOURCE}'  |  "
    f"obs-frame grid: {N_PIXELS_OBS} pix ({wave_obs_grid[0]:.0f}–{wave_obs_grid[-1]:.0f} Å)  |  "
    f"rest-frame grid: {N_PIXELS_REST} pix ({WAVE_REST_MIN:.0f}–{WAVE_REST_MAX:.0f} Å)"
)
```

## Cell 2 — SQL Query Builder

This query selects SDSS Main Galaxy Sample (MGS) spectra with the same cuts
as Jespersen et al. (2025):

| Cut | Value | Reason |
|---|---|---|
| `class = 'GALAXY'` | — | Remove stars and QSOs |
| `z BETWEEN 0.01 AND 0.5` | — | Well-measured redshifts; avoid confusion |
| `zWarning = 0` | — | SDSS flag: clean, unflagged redshift |
| `petroMag_r < 17.8` | AB mag | MGS bright limit (Strauss et al. 2002) |

**Paste the printed SQL into [CasJobs](https://skyserver.sdss.org/casjobs/) for the full 510k dataset.**


```python
def build_sdss_sql(n=None, z_max=0.5):
    """
    Build the SDSS SkyServer SQL query for galaxy spectral metadata.

    The join chain is:
        SpecObj  (spectroscopic table) →
        PhotoObj (photometric table, via bestobjid = objid) →
    yielding plate/mjd/fiberid (to download FITS spectra) plus basic photometry.

    Parameters
    ----------
    n : int or None
        If given, adds TOP n to limit results (for prototyping).
    z_max : float
        Maximum redshift cut.

    Returns
    -------
    str
        A complete SQL string.
    """
    top = f"TOP {n}" if n is not None else ""
    return f"""
SELECT {top}
    s.specobjid,
    s.plate,
    s.mjd,
    s.fiberid,
    s.z           AS redshift,
    s.zwarning,
    s.class       AS specclass,
    s.subclass    AS specsubclass,
    p.petroMag_r
FROM
    SpecObj AS s
JOIN
    PhotoObj AS p ON s.bestobjid = p.objid
WHERE
    s.class    = 'GALAXY'
    AND s.z    BETWEEN 0.01 AND {z_max}
    AND s.zwarning = 0
    AND p.petroMag_r BETWEEN 14.0 AND 17.8
    AND p.petroMag_r != -9999
ORDER BY
    s.plate, s.mjd, s.fiberid
""".strip()


print("SQL for CasJobs (remove TOP for the full dataset):")
print("=" * 62)
print(build_sdss_sql(n=None, z_max=Z_MAX))
print("=" * 62)
```

## Cell 4 — Load Galaxy Metadata

```python
def load_via_astroquery(n, z_max):
    """Live query to SDSS SkyServer. Good for prototyping up to ~10k objects."""
    sql = build_sdss_sql(n=n, z_max=z_max)
    log.info(f"Sending astroquery SQL (TOP {n}) ...")

    for dr in [18, 17, 16]:
        log.info(f"Trying DR{dr} ...")
        try:
            result = SDSS.query_sql(sql, timeout=300, data_release=dr)
            if result is not None and len(result) > 0:
                log.info(f"✓ DR{dr} worked: {len(result):,} rows")
                return result.to_pandas()
        except Exception as e:
            log.warning(f"DR{dr} failed: {type(e).__name__}: {e}")

    # Fallback: direct HTTP request
    log.info("astroquery failed. Trying direct HTTP ...")
    for dr in [18, 17, 16]:
        url = f"https://skyserver.sdss.org/dr{dr}/SkyServerWS/SearchTools/SqlSearch"
        try:
            resp = requests.get(url, params={"cmd": sql, "format": "csv"}, timeout=300)
            text = resp.text.strip()
            if "error" in text[:200].lower() or "<html" in text[:200].lower():
                continue
            df = pd.read_csv(io.StringIO(text))
            if len(df) > 0:
                log.info(f"✓ Direct HTTP DR{dr}: {len(df):,} rows")
                return df
        except Exception:
            continue

    raise RuntimeError(
        "All query methods failed.\n"
        "  Option 1: Paste the Cell 2 SQL into CasJobs (https://skyserver.sdss.org/casjobs/)\n"
        "  Option 2: Set DATA_SOURCE='csv' and point CSV_PATH to the downloaded file."
    )


def load_via_csv(csv_path, n=None):
    """
    Load pre-downloaded CasJobs CSV.
    The full 510k dataset is ~60 MB and loads in < 1 second.
    """
    csv_path = Path(csv_path)
    if not csv_path.exists():
        raise FileNotFoundError(
            f"CSV not found: {csv_path}\n"
            f"Download it from CasJobs (see Cell 2 for SQL)."
        )
    log.info(f"Loading {csv_path} ({csv_path.stat().st_size/1e6:.1f} MB)")
    df = pd.read_csv(csv_path)
    df.columns = df.columns.str.lower().str.strip()
    if n is not None and len(df) > n:
        df = df.head(n)
    log.info(f"Loaded {len(df):,} rows")
    return df


if DATA_SOURCE == "astroquery":
    import requests, io
    df_meta = load_via_astroquery(N_SPECTRA, Z_MAX)
elif DATA_SOURCE == "csv":
    df_meta = load_via_csv(CSV_PATH, n=N_SPECTRA)
else:
    raise ValueError(f"Unknown DATA_SOURCE='{DATA_SOURCE}'")

df_meta.columns = df_meta.columns.str.lower().str.strip()

print(f"\nMetadata: {df_meta.shape[0]:,} rows × {df_meta.shape[1]} columns")
print(df_meta.head(3).to_string())
```

## Cell 5 — Download Spectra

Each SDSS spectrum FITS file contains:

| Column | Meaning |
|---|---|
| `loglam` | log₁₀(wavelength/Å), log-spaced at 10⁻⁴ dex/pixel |
| `flux` | Flux density in 10⁻¹⁷ erg/s/cm²/Å |
| `ivar` | Inverse variance (0 = bad pixel, masked by instrument) |
| `and_mask` | Bitmask: any bit set = some data quality issue |

The SDSS fibre diameter is 3 arcseconds, so the spectrum captures the light
from the central region of each galaxy (not the total flux). WISE photometry
uses profile-fit magnitudes (more like total flux). This aperture mismatch
is discussed in §4.2 of the paper; the MLP learns to account for it empirically.


```python
def download_spectra(df):
    """
    Download raw SDSS spectra via astroquery.

    Returns a list of dicts, one per successfully downloaded spectrum,
    with fields: specobjid, redshift, petroMag_r, wavelength (obs, Å),
    flux (10^-17 erg/s/cm^2/Å), ivar, andmask.
    """
    spectra_raw = []
    n_failed = 0

    for i, row in df.iterrows():
        try:
            fits_list = SDSS.get_spectra(
                plate=int(row["plate"]),
                mjd=int(row["mjd"]),
                fiberID=int(row["fiberid"]),
            )
            if not fits_list:
                n_failed += 1
                continue

            data = fits_list[0][1].data
            spectra_raw.append({
                "specobjid"   : int(row["specobjid"]),
                "redshift"    : float(row["redshift"]),
                "specsubclass": str(row.get("specsubclass", "")).strip(),
                "petroMag_r"  : float(row["petromag_r"]),
                # SDSS wavelengths are stored as log10(lambda/Å)
                "loglam"      : data["loglam"].astype(np.float64),     # log10(λ_obs)
                "wavelength"  : (10.0 ** data["loglam"]).astype(np.float64),  # λ_obs in Å
                "flux"        : data["flux"].astype(np.float64),       # 10^-17 erg/s/cm^2/Å
                "ivar"        : data["ivar"].astype(np.float64),       # inverse variance
                "andmask"     : data["and_mask"].astype(np.int32),     # bitmask
            })
        except Exception as e:
            n_failed += 1
            if n_failed <= 3:
                log.warning(f"  Row {i}: {type(e).__name__}: {e}")

        if (len(spectra_raw) + n_failed) % 50 == 0:
            log.info(f"  {len(spectra_raw)+n_failed}/{len(df)} downloaded "
                     f"({len(spectra_raw)} OK, {n_failed} failed)")

    log.info(f"Download complete: {len(spectra_raw):,} OK, {n_failed} failed")
    return spectra_raw


t0 = time.time()
spectra_raw = download_spectra(df_meta)
log.info(f"Download took {time.time()-t0:.1f}s")
```

## Cell 6 — Preprocessing Pipeline

This cell applies seven sequential steps to produce both formats from each raw spectrum.

### The two output formats in detail

**Observed-frame (for spender):**
- Keep spectra in the *observer* frame (wavelengths NOT divided by 1+z)
- Mask only hardware-flagged bad pixels (ivar ≤ 0, andmask ≠ 0)
- Interpolate to spender's fixed 3921-pixel log-λ grid
- Fill any gaps with zeros (spender is trained to handle this)
- Do NOT normalise — spender's CNN handles all normalisation internally

**Rest-frame normalised (for PCA / clustering):**
- De-redshift: λ_rest = λ_obs / (1 + z)
- Additionally mask skylines in the rest frame
- Interpolate to the 3600-pixel linear rest-frame grid
- Divide by median flux in 5300–5850 Å window
- Clip outliers at ±10 nMAD to suppress cosmic rays

### Why separate treatment?

A key insight about spender: its encoder was **trained on un-normalised observed-frame SDSS spectra**
with sky-line residuals present. Making it robust to those artefacts was an explicit design goal
(see Appendix D of the paper). If we feed it normalised rest-frame spectra, the CNN's learned
convolutional filters — designed for the SDSS log-λ pixel scale in the observer frame — will not
function as intended, and the latent vectors will be wrong.

Mathematically: the spender encoder is a function
    f: ℝ^{3921} → ℝ^6
trained on a specific distribution of inputs. Feeding it out-of-distribution data (wrong
wavelength scale, wrong normalisation) produces meaningless latent vectors.


```python
def preprocess_spectrum(raw_spec, verbose=False):
    """
    Process one raw SDSS spectrum into all three output formats.

    Parameters
    ----------
    raw_spec : dict
        As returned by download_spectra(). Must contain keys:
        wavelength, flux, ivar, andmask, redshift.
    verbose : bool
        Print step-by-step debug info (useful for diagnosing failures).

    Returns
    -------
    dict or None
        None if the spectrum fails quality cuts.
        Otherwise a dict with keys:
            flux_obs         (np.float32, 3921)   — observed frame for spender
            flux_rest_unnorm (np.float32, 3600)   — rest frame, NOT normalised
            flux_rest_norm   (np.float32, 3600)   — rest frame, normalised
            norm_const       (np.float32, scalar) — median flux 5300–5850 Å
    """
    z          = raw_spec["redshift"]
    wave_obs   = raw_spec["wavelength"]   # λ_obs in Å (observed frame)
    flux       = raw_spec["flux"].copy()
    ivar       = raw_spec["ivar"]
    andmask    = raw_spec["andmask"]

    # ─────────────────────────────────────────────────────────────────
    # STEP 1 — Hardware bad-pixel mask
    # Pixels with ivar ≤ 0 are unreliable (read-noise dominated or
    # saturated). andmask bits encode various reduction failures.
    # ─────────────────────────────────────────────────────────────────
    bad_hw = (andmask != 0) | (ivar <= 0) | (~np.isfinite(flux))
    flux[bad_hw] = np.nan   # replace with NaN so interp ignores them

    n_valid = np.isfinite(flux).sum()
    if verbose:
        print(f"  Step 1: bad_hw pixels={bad_hw.sum()}, valid={n_valid}")
    if n_valid < 100:
        return None   # not enough good pixels to work with

    # ─────────────────────────────────────────────────────────────────
    # STEP 2 — Interpolate to spender's observed-frame grid
    # spender's SDSS Instrument defines wave_obs = 10^arange(3.578, 3.97, 0.0001)
    # We need to put our spectrum on this exact grid. We interpolate only from
    # the range where we have valid data; outside that range we fill with 0.
    # (zeros, not NaN, because PyTorch cannot NaN-propagate through a CNN)
    # ─────────────────────────────────────────────────────────────────
    valid_mask = np.isfinite(flux)
    if valid_mask.sum() < 50:
        return None

    try:
        f_obs = interp1d(
            wave_obs[valid_mask], flux[valid_mask],
            kind="linear",
            bounds_error=False,
            fill_value=0.0,   # 0 outside coverage (not NaN — spender expects float tensor)
        )
        flux_obs = f_obs(wave_obs_grid).astype(np.float32)   # shape (3921,)
    except Exception as e:
        if verbose: print(f"  Step 2 failed: {e}")
        return None

    # ─────────────────────────────────────────────────────────────────
    # STEP 3 — De-redshift to rest frame
    # The cosmological redshift stretches wavelengths by a factor (1+z).
    # λ_rest = λ_obs / (1 + z) undoes this stretch.
    # This is a pure rescaling; the flux density itself is not corrected
    # here (a "flux-calibrated" correction is not needed for our purposes).
    # ─────────────────────────────────────────────────────────────────
    wave_rest_native = wave_obs / (1.0 + z)    # λ_rest for each native pixel

    # For the rest-frame grids, ADDITIONALLY mask skylines.
    # Skylines are at fixed *observed* wavelengths λ_sky_obs.
    # In the rest frame they appear at λ_sky_rest = λ_sky_obs / (1+z),
    # but since we already converted to rest-frame coordinates, we
    # just mask the native rest-frame wavelengths that correspond to
    # the known sky-line observer wavelengths.
    bad_sky = bad_hw.copy()
    for sky_lam_obs in SKYLINES_OBS:
        sky_lam_rest = sky_lam_obs / (1.0 + z)
        bad_sky |= (np.abs(wave_rest_native - sky_lam_rest) < SKYLINE_WIDTH_AA)

    flux_for_rest = flux.copy()
    flux_for_rest[bad_sky] = np.nan

    valid_rest = np.isfinite(flux_for_rest)
    if valid_rest.sum() < 100:
        return None

    try:
        f_rest = interp1d(
            wave_rest_native[valid_rest], flux_for_rest[valid_rest],
            kind="linear",
            bounds_error=False,
            fill_value=0.0,
        )
        flux_rest_unnorm = f_rest(wave_rest_grid)   # shape (3600,), NOT normalised
    except Exception:
        return None

    # ─────────────────────────────────────────────────────────────────
    # STEP 4 — Compute the normalisation constant
    # Paper §3.1: "The normalization is a scalar, defined as the median
    # flux in the 5300–5850 Å range."
    # This is the 7th of the 8 MLP input features.
    # It is computed BEFORE normalising, so it carries the true flux level.
    # ─────────────────────────────────────────────────────────────────
    norm_mask = (wave_rest_grid >= NORM_WAVE_MIN) & (wave_rest_grid <= NORM_WAVE_MAX)
    norm_vals = flux_rest_unnorm[norm_mask]
    norm_vals = norm_vals[np.isfinite(norm_vals) & (norm_vals > 0)]
    if len(norm_vals) < 5:
        return None
    norm_const = float(np.median(norm_vals))
    if not np.isfinite(norm_const) or norm_const <= 0:
        return None

    # ─────────────────────────────────────────────────────────────────
    # STEP 5 — Normalise the rest-frame spectrum
    # Dividing by norm_const makes the continuum near 1.0, which helps
    # PCA and k-means focus on spectral *shape* rather than brightness.
    # ─────────────────────────────────────────────────────────────────
    flux_rest_norm = flux_rest_unnorm / norm_const    # dimensionless, ≈ 1.0 continuum

    # ─────────────────────────────────────────────────────────────────
    # STEP 6 — Outlier clipping (rest-frame normalised only)
    # Cosmic rays and reduction artefacts produce isolated high-flux pixels.
    # We clip at ±10 × nMAD (normalised MAD):
    #     nMAD = 1.4826 × median(|x_i − median(x)|)
    # The factor 1.4826 makes nMAD consistent with σ for a Gaussian distribution.
    # nMAD is preferred over std because it is resistant to the same outliers
    # we are trying to remove (std would be inflated by them).
    # ─────────────────────────────────────────────────────────────────
    fm = np.isfinite(flux_rest_norm) & (flux_rest_norm != 0)
    if fm.sum() > 10:
        med  = np.median(flux_rest_norm[fm])
        nmad = 1.4826 * np.median(np.abs(flux_rest_norm[fm] - med))
        flux_rest_norm = np.clip(flux_rest_norm, med - 10 * nmad, med + 10 * nmad)

    return {
        "flux_obs"         : flux_obs.astype(np.float32),
        "flux_rest_unnorm" : flux_rest_unnorm.astype(np.float32),
        "flux_rest_norm"   : flux_rest_norm.astype(np.float32),
        "norm_const"       : np.float32(norm_const),
    }
```

```python
# ── Apply preprocessing to all downloaded spectra ─────────────────────
obs_rows, unnorm_rows, norm_rows, norm_const_list, meta_clean = [], [], [], [], []

n_failed = 0
t0 = time.time()

for i, raw_spec in enumerate(spectra_raw):
    result = preprocess_spectrum(raw_spec)
    if result is None:
        n_failed += 1
        continue

    obs_rows.append(result["flux_obs"])
    unnorm_rows.append(result["flux_rest_unnorm"])
    norm_rows.append(result["flux_rest_norm"])
    norm_const_list.append(result["norm_const"])

    meta_clean.append({
        "specobjid"   : raw_spec["specobjid"],
        "redshift"    : raw_spec["redshift"],
        "specsubclass": raw_spec["specsubclass"],
        "petroMag_r"  : raw_spec["petroMag_r"],
    })

    if (i + 1) % 50 == 0:
        log.info(f"  Preprocessed {i+1}/{len(spectra_raw)}  "
                 f"({len(norm_rows)} OK, {n_failed} failed)")

# ── Stack into numpy arrays ────────────────────────────────────────────
# np.vstack turns a list of (n_pix,) arrays into (N, n_pix)
X_obs         = np.vstack(obs_rows)           # (N, 3921)
X_rest_unnorm = np.vstack(unnorm_rows)        # (N, 3600)
X_rest_norm   = np.vstack(norm_rows)          # (N, 3600)  ← the "spectra_matrix" from before
norm_consts   = np.array(norm_const_list, dtype=np.float32)  # (N,)

redshifts = np.array([m["redshift"]   for m in meta_clean], dtype=np.float32)
specobjids = np.array([m["specobjid"] for m in meta_clean], dtype=np.int64)

N = len(meta_clean)
log.info(f"Preprocessing complete in {time.time()-t0:.1f}s: {N:,} OK, {n_failed} failed")
log.info(f"Pass rate: {N}/{len(spectra_raw)} ({100*N/max(1,len(spectra_raw)):.1f}%)")
print(f"\nX_obs shape:         {X_obs.shape}  (observed-frame, for spender)")
print(f"X_rest_unnorm shape: {X_rest_unnorm.shape}  (rest-frame, unnormalised)")
print(f"X_rest_norm shape:   {X_rest_norm.shape}  (rest-frame, normalised)")
print(f"norm_consts shape:   {norm_consts.shape}  (scalar per galaxy)")
print(f"\nnorm_const stats: mean={norm_consts.mean():.2f}, std={norm_consts.std():.2f}, "
      f"range=[{norm_consts.min():.2f}, {norm_consts.max():.2f}]")
```

## Cell 7 — Diagnostic Plots

```python
# ── Known spectral lines to annotate on rest-frame plots ──────────────
lines = {
    "[OII]": 3727, "Hβ": 4861, "[OIII]": 5007,
    "MgI b": 5175, "NaI D": 5893, "Hα": 6563, "[SII]": 6717,
}

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

# ── Panel 1: Redshift distribution ────────────────────────────────────
ax = axes[0, 0]
ax.hist(redshifts, bins=40, color="steelblue", edgecolor="white")
ax.set_xlabel("Spectroscopic redshift z")
ax.set_ylabel("Count")
ax.set_title("Redshift distribution")

# ── Panel 2: norm_const distribution ──────────────────────────────────
# The norm_const is the median flux in the 5300–5850 Å V-band window,
# in units of 10^{-17} erg/s/cm²/Å. It spans ~3 decades because SDSS
# galaxies span a wide range of apparent brightnesses.
ax = axes[0, 1]
ax.hist(np.log10(norm_consts), bins=40, color="darkorange", edgecolor="white")
ax.set_xlabel("log₁₀(norm_const) [10⁻¹⁷ erg/s/cm²/Å]")
ax.set_ylabel("Count")
ax.set_title("Normalisation constant distribution")
ax.axvline(np.log10(np.median(norm_consts)), color="red", lw=2,
           label=f"median={np.median(norm_consts):.1f}")
ax.legend()

# ── Panel 3: Mean normalised rest-frame spectrum ───────────────────────
ax = axes[0, 2]
mu  = X_rest_norm.mean(axis=0)
p16 = np.percentile(X_rest_norm, 16, axis=0)
p84 = np.percentile(X_rest_norm, 84, axis=0)
ax.plot(wave_rest_grid, mu, "steelblue", lw=1.2, label="Mean")
ax.fill_between(wave_rest_grid, p16, p84, color="steelblue", alpha=0.2, label="16–84th %ile")
ax.axvspan(NORM_WAVE_MIN, NORM_WAVE_MAX, color="gold", alpha=0.2, label="Norm window")
for name, lam in lines.items():
    if WAVE_REST_MIN < lam < WAVE_REST_MAX:
        ax.axvline(lam, ls=":", lw=0.7, color="gray", alpha=0.6)
ax.set_xlabel("Rest-frame λ (Å)")
ax.set_ylabel("Normalised flux")
ax.set_title("Mean normalised rest-frame spectrum")
ax.legend(fontsize=8)
ax.set_xlim(WAVE_REST_MIN, WAVE_REST_MAX)

# ── Panel 4: Observed-frame spectra (what goes into spender) ──────────
ax = axes[1, 0]
mu_obs  = X_obs.mean(axis=0)
p16_obs = np.percentile(X_obs, 16, axis=0)
p84_obs = np.percentile(X_obs, 84, axis=0)
ax.plot(wave_obs_grid, mu_obs, "darkorange", lw=1.2, label="Mean")
ax.fill_between(wave_obs_grid, p16_obs, p84_obs, color="darkorange", alpha=0.2)
ax.set_xlabel("Observed-frame λ (Å)")
ax.set_ylabel("Flux [10⁻¹⁷ erg/s/cm²/Å]")
ax.set_title("Mean observed-frame spectrum (spender input)")
ax.legend(fontsize=8)

# ── Panel 5: Side-by-side: single galaxy in both frames ───────────────
# Picking a galaxy at z ≈ 0.1 shows how the spectral features shift.
mid_z_idx = np.argmin(np.abs(redshifts - 0.1))
ax = axes[1, 1]
ax.plot(wave_rest_grid, X_rest_norm[mid_z_idx], "steelblue", lw=0.8, label="Rest-frame norm.")
ax2 = ax.twinx()
ax2.plot(wave_obs_grid, X_obs[mid_z_idx], "darkorange", lw=0.8, alpha=0.7, label="Observed-frame")
ax.set_xlabel("Rest-frame λ (Å) [blue axis]")
ax.set_ylabel("Normalised flux", color="steelblue")
ax2.set_ylabel("Flux [10⁻¹⁷ erg/s/cm²/Å]", color="darkorange")
ax.set_title(f"Galaxy at z = {redshifts[mid_z_idx]:.3f}")
lines1, labels1 = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax.legend(lines1 + lines2, labels1 + labels2, fontsize=8)

# ── Panel 6: norm_const vs petroMag_r ─────────────────────────────────
# Since both are proxies for galaxy brightness, we expect a tight correlation.
# Scatter shows the colour and redshift dependence.
ax = axes[1, 2]
pmr = np.array([m["petroMag_r"] for m in meta_clean], dtype=np.float32)
sc = ax.scatter(pmr, np.log10(norm_consts), c=redshifts, s=8, alpha=0.5, cmap="plasma")
plt.colorbar(sc, ax=ax, label="Redshift z")
ax.set_xlabel("petroMag_r (AB mag)")
ax.set_ylabel("log₁₀(norm_const)")
ax.set_title("Brightness vs. norm_const (coloured by z)")

plt.suptitle("SDSS Preprocessing Diagnostics", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig(OUT_DIR / "sdss_preprocessing_diagnostics.png", dpi=150, bbox_inches="tight")
plt.show()
log.info(f"Saved diagnostics to {OUT_DIR}/sdss_preprocessing_diagnostics.png")
```

```python
# Reload every saved file and check shapes / dtypes.
print("═" * 60)
print("RELOAD VERIFICATION")
print("═" * 60)

checks = {
    "spectra_obs_frame.npy":   ((N, N_PIXELS_OBS),  np.float32),
    "spectra_restframe_norm.npy": ((N, N_PIXELS_REST), np.float32),
    "spectra_restframe_unnorm.npy": ((N, N_PIXELS_REST), np.float32),
    "norm_consts.npy":         ((N,),               np.float32),
    "wave_obs_grid.npy":       ((N_PIXELS_OBS,),    np.float64),
    "wave_grid.npy":           ((N_PIXELS_REST,),   np.float64),
}

all_ok = True
for fname, (expected_shape, expected_dtype) in checks.items():
    arr = np.load(OUT_DIR / fname)
    ok_shape = arr.shape == expected_shape
    ok_dtype = arr.dtype == expected_dtype
    status = "✓" if (ok_shape and ok_dtype) else "✗"
    if not (ok_shape and ok_dtype):
        all_ok = False
    print(f"  {status} {fname:<45s}  shape={str(arr.shape):<15s}  dtype={arr.dtype}")
    if not ok_shape:
        print(f"      ⚠️  Expected shape {expected_shape}, got {arr.shape}")
    if not ok_dtype:
        print(f"      ⚠️  Expected dtype {expected_dtype}, got {arr.dtype}")

# Spot-check: norm_consts should be finite and positive
nc = np.load(OUT_DIR / "norm_consts.npy")
n_bad_nc = (~np.isfinite(nc) | (nc <= 0)).sum()
print(f"\n  {'✓' if n_bad_nc==0 else '✗'} norm_consts: {n_bad_nc} non-positive/non-finite")

# Spot-check alignment: obs-frame spectrum and rest-frame spectrum should be same galaxy
obs  = np.load(OUT_DIR / "spectra_obs_frame.npy")
norm = np.load(OUT_DIR / "spectra_restframe_norm.npy")
meta = np.load(OUT_DIR / "metadata.npy")
print(f"\nSpot check (first 3 galaxies):")
for i in range(min(3, N)):
    print(f"  [{i}] z={meta['redshift'][i]:.4f}  "
          f"obs_max={obs[i].max():.2f}  "
          f"norm_median={np.median(norm[i]):.3f}  "
          f"norm_const={nc[i]:.3f}")

print(f"\n{'✓ All checks passed!' if all_ok else '✗ Some checks failed — see above.'}")
```
