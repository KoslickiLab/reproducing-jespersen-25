# Reproducing Figure 3: PP-Plots for Optical → IR Prediction
### Jespersen et al. (2025) — *"The optical and infrared are connected"*

---

**Pipeline:**
1. Load preprocessed SDSS spectra + WISE photometry
2. Encode spectra → 6-D latents (via `spender`) + compute norm constants
3. Train MLP: [6 latents, norm_const, redshift] → [W1, W2, W3, W4] Vega mags
4. Compute $\hat{p}$ (Eq. 2–3) and plot the PP-plot (Figure 3)

**Prerequisites:** Run these first:
- `sdss_spectra_clustering.ipynb` → `sdss_output/`
- `wise_data_preprocessing.ipynb` → `wise_output/`

---

### What is a PP-plot?

For each galaxy, compute $\hat{p}$ = probability that the observed WISE mag is consistent
with the predicted mag, given observational uncertainty:

$$\hat{p} = \int p_{\text{data}}(\text{WISE}) \cdot p_{\text{model}}(\text{WISE}|\text{optical})\,d\text{WISE}$$

For a point prediction (our MLP), this becomes $p_{\text{data}}(\hat{y})$ —
evaluate the data Gaussian at the prediction. If the model is perfect, $\hat{p}$
is uniform on [0,1] and the CDF is the diagonal.

### MLP architecture (paper §3.1)

```
[latent₁..₆, norm_const, z]  →  Linear(8→20)+ReLU  →  Linear(20→20)+ReLU  →  Linear(20→4)×norm
```
2 hidden layers × 20 units = 3 weight matrices = **684 parameters**.
Loss: Gaussian NLL using WISE errors as σ (non-detections get σ = 10⁴).

````python
import warnings; warnings.filterwarnings("ignore")
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from scipy.stats import norm as scipy_norm
import time, sys, logging

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

logging.basicConfig(level=logging.INFO, format="%(asctime)s [%(levelname)s] %(message)s", datefmt="%H:%M:%S")
log = logging.getLogger("PPplot")

%matplotlib inline
plt.rcParams.update({"figure.dpi": 120, "font.size": 11})

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"PyTorch {torch.__version__}  |  Device: {DEVICE}")
if DEVICE.type == "cuda":
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    print(f"  VRAM: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
print("All imports OK ✓")
````

## Cell 1 — Configuration

````python
SDSS_DIR = Path("sdss_output")
WISE_DIR = Path("wise_output")
OUT_DIR  = Path("ppplot_output"); OUT_DIR.mkdir(exist_ok=True)

NORM_WAVE_MIN, NORM_WAVE_MAX = 5300.0, 5850.0

# Train/test split (paper: 70/10/20; we simplify to 80/20 for prototype)
TEST_FRACTION = 0.20

# MLP hyperparams (paper §3.1)
HIDDEN_SIZE   = 20
N_LAYERS      = 2      # 2 hidden layers → 3 weight matrices → 684 params
LEARNING_RATE = 1e-3
N_EPOCHS      = 300
BATCH_SIZE    = 64     # scale up to 4096 for 510k dataset

# ═══════════════════════════════════════════════════════════════════════
# spender encoding batch size — tune for your GPU VRAM
#   200 objects:  SPENDER_BATCH = 256  (any device)
#   510k objects: SPENDER_BATCH = 10000 (H200 w/ 141 GB)
# ═══════════════════════════════════════════════════════════════════════
SPENDER_BATCH = 256

log.info(f"Config ready. Device={DEVICE}")
````

## Cell 2 — Load preprocessed data

````python
# ── SDSS ──────────────────────────────────────────────────────────────
X_spectra = np.load(SDSS_DIR / "spectra_matrix.npy")    # (N_sdss, 3600)
wave_grid = np.load(SDSS_DIR / "wave_grid.npy")          # (3600,)
sdss_meta = np.load(SDSS_DIR / "metadata.npy")           # structured

log.info(f"SDSS: {X_spectra.shape[0]:,} spectra, {X_spectra.shape[1]} pixels")
log.info(f"  wavelength: {wave_grid[0]:.0f}–{wave_grid[-1]:.0f} Å")
log.info(f"  flux range: [{X_spectra.min():.3f}, {X_spectra.max():.3f}]")

# ── WISE ──────────────────────────────────────────────────────────────
wise = np.load(WISE_DIR / "wise_photometry.npz")
log.info(f"WISE: {wise['mag'].shape[0]:,} galaxies, keys={list(wise.keys())}")

# ── Align arrays ─────────────────────────────────────────────────────
idx_path = WISE_DIR / "wise_matched_sdss_indices.npy"
if idx_path.exists():
    sdss_idx = np.load(idx_path)
    X_spec = X_spectra[sdss_idx]
    z_arr  = sdss_meta["redshift"][sdss_idx]
    wise_mag     = wise["mag"]
    wise_mag_err = wise["mag_err"]
    wise_det     = wise["detected"]
    log.info(f"Aligned via saved indices: {len(sdss_idx):,} matched galaxies")
else:
    # Fallback: match by specobjid
    log.info("Matching by specobjid ...")
    sdss_ids = sdss_meta["specobjid"]
    wise_ids = wise["specobjid"]
    wlookup = {}; 
    for i,wid in enumerate(wise_ids): wlookup.setdefault(int(wid), i)
    sk, wk = [], []
    for si, sid in enumerate(sdss_ids):
        if int(sid) in wlookup:
            sk.append(si); wk.append(wlookup[int(sid)])
    sk, wk = np.array(sk), np.array(wk)
    X_spec       = X_spectra[sk]
    z_arr        = sdss_meta["redshift"][sk]
    wise_mag     = wise["mag"][wk]
    wise_mag_err = wise["mag_err"][wk]
    wise_det     = wise["detected"][wk]
    log.info(f"Matched: {len(sk):,} galaxies")

N = len(X_spec)
log.info(f"Joint dataset: {N:,} galaxies")
````

## Cell 3 — Diagnostic: loaded data peek

````python
print("═" * 60)
print("LOADED DATA PEEK")
print("═" * 60)
print(f"Spectra:    shape={X_spec.shape}  dtype={X_spec.dtype}")
print(f"Redshifts:  range=[{z_arr.min():.4f}, {z_arr.max():.4f}]  NaN={np.isnan(z_arr).sum()}")
print(f"WISE mag:   shape={wise_mag.shape}")
print(f"WISE err:   shape={wise_mag_err.shape}")

# Check for degenerate spectra
all_zero = (X_spec == 0).all(axis=1).sum()
all_nan  = np.isnan(X_spec).all(axis=1).sum()
has_nan  = np.isnan(X_spec).any(axis=1).sum()
print(f"\nSpectra issues:")
print(f"  All-zero rows:  {all_zero}")
print(f"  All-NaN rows:   {all_nan}")
print(f"  Any-NaN rows:   {has_nan}")

# Detection counts
for j, b in enumerate(["W1","W2","W3","W4"]):
    n_det = wise_det[:,j].sum()
    print(f"  {b} detections: {n_det:,} / {N:,} ({100*n_det/N:.1f}%)")

# Spot-check alignment: first 3 galaxies
print(f"\nSpot check (first 3):")
for i in range(min(3, N)):
    z = z_arr[i]
    flux_med = np.nanmedian(X_spec[i])
    w1 = wise_mag[i, 0]
    w1d = wise_det[i, 0]
    print(f"  [{i}] z={z:.4f}  spec_median={flux_med:.3f}  W1={w1:.2f} det={w1d}")
````

## Cell 4 — Encode spectra with `spender`

`spender` (Melchior et al. 2023) is a convolutional autoencoder trained on the full SDSS galaxy sample. It compresses each 3600-pixel spectrum into **6 latent variables** + a **normalisation constant**.

The latent representation is redshift-invariant (Liang et al. 2023b added loss terms to enforce this), which means the 6 numbers capture intrinsic spectral diversity: stellar populations, emission-line strengths, dust content, etc.

**Important subtlety:** The spectra from the SDSS notebook are already normalised (divided by the median flux in 5300–5850 Å). `spender` handles normalisation internally, so feeding it already-normalised spectra means its returned `norm_const` will be ≈ 1.0 rather than the true flux level. For a faithful reproduction at full scale, you should feed `spender` the **un-normalised** spectra. The code below handles both cases.

````python
import spender

log.info("Loading spender model (SDSS-main) ...")
t0 = time.time()
model_sp, _ = spender.load("SDSS-main")
model_sp.eval()
log.info(f"spender loaded in {time.time()-t0:.1f}s")

# Move to GPU if available (the spectra matrix may be large)
spender_device = DEVICE
# For 510k spectra: ~7.3 GB float32 → fits on H200 (141 GB)
# For prototype: fits anywhere
model_sp = model_sp.to(spender_device)

wave_t = torch.tensor(wave_grid, dtype=torch.float32).to(spender_device)

# ── Encode in batches ────────────────────────────────────────────────
latents_list = []
norms_list   = []
n_batches = (N + SPENDER_BATCH - 1) // SPENDER_BATCH

log.info(f"Encoding {N:,} spectra in {n_batches} batches of {SPENDER_BATCH} ...")
t0 = time.time()

with torch.no_grad():
    for bi in range(n_batches):
        i0 = bi * SPENDER_BATCH
        i1 = min(i0 + SPENDER_BATCH, N)
        
        batch = torch.tensor(
            np.nan_to_num(X_spec[i0:i1], nan=0.0),
            dtype=torch.float32
        ).to(spender_device)
        
        lat, nc = model_sp.encode(batch, wave_t)
        latents_list.append(lat.cpu().numpy())
        norms_list.append(nc.cpu().numpy().flatten())
        
        if (bi + 1) % max(1, n_batches // 5) == 0 or bi == n_batches - 1:
            elapsed = time.time() - t0
            rate = (i1) / elapsed
            log.info(f"  batch {bi+1}/{n_batches}  ({i1:,}/{N:,})  {rate:.0f} spec/s")

latents     = np.concatenate(latents_list, axis=0)   # (N, 6)
norm_consts = np.concatenate(norms_list, axis=0)     # (N,)

total_time = time.time() - t0
log.info(f"Encoding complete: {total_time:.1f}s  ({N/total_time:.0f} spec/s)")
log.info(f"Latent shape: {latents.shape}  |  Norm consts shape: {norm_consts.shape}")
````

## Cell 5 — Diagnostic: spender latent space

````python
print("═" * 60)
print("SPENDER ENCODING DIAGNOSTICS")
print("═" * 60)

# Latent statistics
print(f"\nLatent variable statistics:")
print(f"  {'dim':<5s} {'mean':>8s} {'std':>8s} {'min':>8s} {'max':>8s}")
print(f"  {'-'*40}")
for d in range(6):
    col = latents[:, d]
    print(f"  L{d+1:<4d} {col.mean():>8.3f} {col.std():>8.3f} {col.min():>8.3f} {col.max():>8.3f}")

# Norm const statistics
print(f"\nNorm constants:")
print(f"  mean={norm_consts.mean():.4f}  std={norm_consts.std():.4f}")
print(f"  range=[{norm_consts.min():.4f}, {norm_consts.max():.4f}]")
if norm_consts.mean() < 0.5 or norm_consts.mean() > 2.0:
    print(f"  ⚠️  Norm consts far from 1.0 — spectra may not be pre-normalised as expected")
else:
    print(f"  ✓ Norm consts ≈ 1 (spectra were pre-normalised)")

# Check for NaN/Inf in latents
n_nan_lat = np.isnan(latents).any(axis=1).sum()
n_inf_lat = np.isinf(latents).any(axis=1).sum()
print(f"\n{'✓' if n_nan_lat==0 else '⚠️'} NaN latents: {n_nan_lat}")
print(f"{'✓' if n_inf_lat==0 else '⚠️'} Inf latents: {n_inf_lat}")

# Quick latent scatter (first 2 dims, coloured by redshift)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
ax = axes[0]
sc = ax.scatter(latents[:, 0], latents[:, 1], c=z_arr, s=5, alpha=0.6, cmap="plasma")
plt.colorbar(sc, ax=ax, label="Redshift")
ax.set_xlabel("Latent 1"); ax.set_ylabel("Latent 2")
ax.set_title("spender latent space (coloured by z)")

ax = axes[1]
ax.hist(norm_consts, bins=40, color="steelblue", edgecolor="white")
ax.set_xlabel("Norm constant"); ax.set_ylabel("Count")
ax.set_title("Normalisation constants from spender")
plt.tight_layout(); plt.show()
````

## Cell 6 — Build MLP inputs & train/test split

````python
# Input: [latent_1..6, norm_const, redshift]
X_input = np.column_stack([
    latents,                            # (N, 6)
    norm_consts.reshape(-1, 1),          # (N, 1)
    z_arr.reshape(-1, 1),                # (N, 1)
]).astype(np.float32)

# Target: [W1, W2, W3, W4] Vega mags (NaN → 0, masked by huge σ)
Y_target = np.nan_to_num(wise_mag, nan=0.0).astype(np.float32)
Y_errors = wise_mag_err.astype(np.float32)

log.info(f"X_input: {X_input.shape}  Y_target: {Y_target.shape}  Y_errors: {Y_errors.shape}")

# Diagnostic: check for NaN in inputs
n_nan_x = np.isnan(X_input).any(axis=1).sum()
if n_nan_x > 0:
    log.warning(f"⚠️  {n_nan_x} input rows contain NaN — replacing with 0")
    X_input = np.nan_to_num(X_input, nan=0.0)

# ── Split ────────────────────────────────────────────────────────────
perm = np.random.permutation(N)
N_test = int(N * TEST_FRACTION)
N_train = N - N_test
idx_train, idx_test = perm[:N_train], perm[N_train:]

X_tr, X_te = X_input[idx_train],  X_input[idx_test]
Y_tr, Y_te = Y_target[idx_train], Y_target[idx_test]
E_tr, E_te = Y_errors[idx_train], Y_errors[idx_test]
det_te     = wise_det[idx_test]

print(f"\nTrain: {N_train:,}  |  Test: {N_test:,}")
for j, b in enumerate(["W1","W2","W3","W4"]):
    print(f"  {b} test detections: {det_te[:,j].sum():,} / {N_test:,}")
````

## Cell 7 — Define MLP + loss

````python
class WISEPredictor(nn.Module):
    """
    Paper's MLP: 2 hidden layers × 20 units, ReLU, 684 params.
    Predicts 4 WISE magnitudes from 8 input features.
    norm_const is applied after the output layer (paper §3.1).
    """
    def __init__(self, n_in=8, n_hid=20, n_layers=2, n_out=4):
        super().__init__()
        layers = [nn.Linear(n_in, n_hid), nn.ReLU()]
        for _ in range(n_layers - 1):
            layers += [nn.Linear(n_hid, n_hid), nn.ReLU()]
        layers.append(nn.Linear(n_hid, n_out))
        self.net = nn.Sequential(*layers)
    
    def forward(self, x):
        raw = self.net(x)              # (batch, 4)
        norm = x[:, 6:7]               # norm_const column
        return raw * norm              # scale by norm_const


def gauss_nll(y_pred, y_true, sigma):
    """Gaussian NLL (paper §3.1). Non-detections have σ=1e4 → ~constant loss."""
    return torch.mean(0.5 * torch.log(sigma) + 0.5 * ((y_true - y_pred) / sigma)**2)


model = WISEPredictor(n_in=8, n_hid=HIDDEN_SIZE, n_layers=N_LAYERS, n_out=4).to(DEVICE)
n_params = sum(p.numel() for p in model.parameters())
print(f"Model parameters: {n_params}  (paper: 684)")
assert n_params == 684, f"Expected 684 params, got {n_params}"
print(f"✓ Parameter count matches paper")
print(f"\n{model}")
````

## Cell 8 — Train

````python
train_ds = TensorDataset(
    torch.tensor(X_tr, dtype=torch.float32),
    torch.tensor(Y_tr, dtype=torch.float32),
    torch.tensor(E_tr, dtype=torch.float32),
)
# Scale batch size for full dataset
effective_batch = min(BATCH_SIZE, N_train)
if N_train > 50000:
    effective_batch = 4096
    log.info(f"Large dataset: using batch_size={effective_batch}")

train_loader = DataLoader(train_ds, batch_size=effective_batch, shuffle=True)
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

losses = []
model.train()
t0 = time.time()

for epoch in range(N_EPOCHS):
    ep_loss = 0.0; nb = 0
    for xb, yb, eb in train_loader:
        xb, yb, eb = xb.to(DEVICE), yb.to(DEVICE), eb.to(DEVICE)
        optimizer.zero_grad()
        loss = gauss_nll(model(xb), yb, eb)
        loss.backward()
        optimizer.step()
        ep_loss += loss.item(); nb += 1
    avg = ep_loss / nb
    losses.append(avg)
    if (epoch+1) % 50 == 0 or epoch == 0:
        log.info(f"Epoch {epoch+1:>4d}/{N_EPOCHS}  loss={avg:.4f}")

total_train = time.time() - t0
log.info(f"Training done: {total_train:.1f}s  final_loss={losses[-1]:.4f}")

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(losses, color="steelblue")
ax.set_xlabel("Epoch"); ax.set_ylabel("Gaussian NLL")
ax.set_title(f"Training curve ({N_train:,} samples, {total_train:.1f}s)")
plt.tight_layout(); plt.show()
````

## Cell 9 — Predict on test set + diagnostics

````python
model.eval()
with torch.no_grad():
    Y_pred = model(torch.tensor(X_te, dtype=torch.float32).to(DEVICE)).cpu().numpy()

log.info(f"Predictions: {Y_pred.shape}")

# ── Reduced χ² per band (paper Table 3) ──────────────────────────────
paper_chi2 = {"W1":1.33, "W2":1.17, "W3":2.23, "W4":1.41}
print(f"\n{'Band':<10s} {'χ²_N':>8s} {'Paper':>8s} {'N_det':>7s} {'Match?':>8s}")
print("-" * 50)
for j, b in enumerate(["W1","W2","W3","W4"]):
    det = det_te[:, j]
    if det.sum() == 0:
        print(f"{b:<10s} {'N/A':>8s} {paper_chi2[b]:>8.2f} {0:>7d}")
        continue
    chi = (Y_te[det, j] - Y_pred[det, j]) / E_te[det, j]
    chi2n = np.mean(chi**2)
    # At small N, χ² will be noisy. Flag if wildly off.
    status = "✓" if chi2n < 10*paper_chi2[b] else "⚠️ high"
    print(f"{b:<10s} {chi2n:>8.2f} {paper_chi2[b]:>8.2f} {det.sum():>7d} {status:>8s}")

# Spot-check: 5 random test galaxies
print(f"\nSpot check (5 random test galaxies):")
rng_idx = np.random.choice(N_test, min(5, N_test), replace=False)
for i in rng_idx:
    z = X_te[i, 7]  # redshift is column 7
    for j, b in enumerate(["W1","W2","W3","W4"]):
        if det_te[i, j]:
            res = (Y_pred[i,j] - Y_te[i,j]) / E_te[i,j]
            print(f"  [{i}] z={z:.3f} {b}: obs={Y_te[i,j]:.2f}±{E_te[i,j]:.3f}  "
                  f"pred={Y_pred[i,j]:.2f}  χ={res:+.2f}")
````

## Cell 10 — Compute PP-plot metric $\hat{p}$ (Eq. 2–3)

For our delta-function posterior (point prediction):
$$\hat{p}_i = \mathcal{N}(\hat{y}_i \,|\, y_i^\text{obs}, \sigma_i) = \frac{1}{\sigma_i \sqrt{2\pi}} \exp\!\left(-\frac{(y_i - \hat{y}_i)^2}{2\sigma_i^2}\right)$$

Data baseline: $\hat{p}_i^\text{data} = \frac{1}{\sigma_i \sqrt{2\pi}}$ (peak of the Gaussian).

````python
phat_emp  = {}  # band → array
phat_data = {}  # band → array

paper_phat = {"W1":0.66, "W2":0.70, "W3":0.63, "W4":0.70}
paper_data = {"W1":0.70, "W2":0.70, "W3":0.70, "W4":0.70}

print(f"{'Band':<6s} {'<p̂>_emp':>10s} {'<p̂>_data':>10s} {'Paper_emp':>10s} {'Paper_data':>10s} {'N':>7s}")
print("-" * 58)

for j, b in enumerate(["W1","W2","W3","W4"]):
    det = det_te[:, j]
    if det.sum() < 2:
        print(f"{b:<6s} {'—':>10s} {'—':>10s} {paper_phat[b]:>10.2f} {paper_data[b]:>10.2f} {det.sum():>7d}")
        continue
    
    y_obs = Y_te[det, j]
    y_prd = Y_pred[det, j]
    sig   = E_te[det, j]
    
    phat_emp[b]  = scipy_norm.pdf(y_prd, loc=y_obs, scale=sig)
    phat_data[b] = scipy_norm.pdf(0, loc=0, scale=sig)  # 1/(σ√2π)
    
    me = np.mean(phat_emp[b])
    md = np.mean(phat_data[b])
    print(f"{b:<6s} {me:>10.4f} {md:>10.4f} {paper_phat[b]:>10.2f} {paper_data[b]:>10.2f} {det.sum():>7d}")

print(f"\nNote: With {N_test} test galaxies, these means will be noisy.")
print(f"The paper uses ~100k test galaxies for the values in Table 4.")
````

## Cell 11 — Figure 3: PP-plot

**How to read it:**
- Diagonal = perfect, unbiased model
- Empirical line close to Data baseline = good predictions
- Line below diagonal = overconfident/biased

````python
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
blabels = {"W1":r"WISE 3.4 $\mu$m", "W2":r"WISE 4.6 $\mu$m",
           "W3":r"WISE 12.1 $\mu$m", "W4":r"WISE 22.2 $\mu$m"}

for idx, b in enumerate(["W1","W2","W3","W4"]):
    ax = axes[idx//2, idx%2]
    
    if b not in phat_emp:
        ax.text(0.5, 0.5, f"No {b} detections in test set",
                transform=ax.transAxes, ha="center", va="center")
        ax.set_title(blabels[b]); continue
    
    pe = np.sort(phat_emp[b])
    pd_ = np.sort(phat_data[b])
    ne, nd = len(pe), len(pd_)
    
    # Build CDFs on a common threshold grid
    thresholds = np.linspace(0, max(pe.max(), pd_.max()), 500)
    cdf_e = np.searchsorted(pe,  thresholds) / ne
    cdf_d = np.searchsorted(pd_, thresholds) / nd
    
    # Diagonal
    ax.plot([0,1],[0,1], 'r-', lw=1.5, label='Perfect model')
    # Data baseline
    ax.plot(cdf_d, cdf_d, 'k--', lw=1.5, alpha=0.7, label='Data')
    # Empirical
    ax.plot(cdf_d, cdf_e, 'g-', lw=2.0, label='Empirical')
    
    ax.annotate('Overconfident\nand biased', xy=(0.65, 0.35), fontsize=8, color='gray',
                ha='center', arrowprops=dict(arrowstyle='->', color='gray', lw=0.8),
                xytext=(0.45, 0.15))
    
    ax.set_xlabel('True CDF'); ax.set_ylabel('Model CDF')
    ax.set_title(blabels[b])
    ax.set_xlim(0,1); ax.set_ylim(0,1); ax.set_aspect('equal')
    ax.legend(loc='upper left', fontsize=8)

plt.suptitle('PP-plot: Empirical model prediction quality\n'
             '(Reproducing Figure 3 of Jespersen et al. 2025)',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(OUT_DIR / "ppplot_figure3.png", dpi=150, bbox_inches="tight")
plt.show()
log.info(f"Saved {OUT_DIR / 'ppplot_figure3.png'}")
````

## Cell 12 — χ-distribution (paper Figure 15 analogue)

## Cell 16 — Full-scale implementation

Concrete, runnable code for scaling to the full 510k dataset on your server.

````python
def run_full_scale(
    sdss_dir="sdss_output",
    wise_dir="wise_output",
    out_dir="ppplot_output_fullscale",
    device="cuda:0",
    spender_batch=10_000,
    mlp_batch=4096,
    n_epochs=300,
    seed=42,
):
    """
    End-to-end full-scale pipeline.
    
    On your 768-core / 3TB RAM / 2×H200 server:
      - spender encoding: ~5-10 min on 1 GPU (510k × 3600 pixels)
      - MLP training:     ~30s on GPU, ~2 min on CPU
      - Total:            ~12 min
    
    Call this function after downloading the full dataset.
    """
    import torch
    import spender as sp
    from pathlib import Path
    
    np.random.seed(seed)
    torch.manual_seed(seed)
    out = Path(out_dir); out.mkdir(exist_ok=True)
    dev = torch.device(device)
    
    # ── Load data ────────────────────────────────────────────────────
    print("[1/5] Loading data ...")
    X_spec  = np.load(f"{sdss_dir}/spectra_matrix.npy")    # (N, 3600)
    wgrid   = np.load(f"{sdss_dir}/wave_grid.npy")          # (3600,)
    meta    = np.load(f"{sdss_dir}/metadata.npy")
    wise    = np.load(f"{wise_dir}/wise_photometry.npz")
    sidx    = np.load(f"{wise_dir}/wise_matched_sdss_indices.npy")
    
    X_spec = X_spec[sidx]
    z_arr  = meta["redshift"][sidx]
    N = len(X_spec)
    print(f"  {N:,} matched galaxies")
    
    # ── Encode with spender ──────────────────────────────────────────
    print("[2/5] Encoding with spender ...")
    model_sp, _ = sp.load("SDSS-main")
    model_sp.eval().to(dev)
    wave_t = torch.tensor(wgrid, dtype=torch.float32).to(dev)
    
    lat_list, nc_list = [], []
    t0 = time.time()
    with torch.no_grad():
        for i in range(0, N, spender_batch):
            batch = torch.tensor(
                np.nan_to_num(X_spec[i:i+spender_batch], nan=0.0),
                dtype=torch.float32
            ).to(dev)
            lat, nc = model_sp.encode(batch, wave_t)
            lat_list.append(lat.cpu().numpy())
            nc_list.append(nc.cpu().numpy().flatten())
            if (i // spender_batch) % 10 == 0:
                print(f"  {min(i+spender_batch, N):,}/{N:,} ...", flush=True)
    
    latents = np.concatenate(lat_list, axis=0)
    norms   = np.concatenate(nc_list, axis=0)
    print(f"  Done in {time.time()-t0:.1f}s")
    
    # ── Build arrays ─────────────────────────────────────────────────
    print("[3/5] Building MLP arrays ...")
    X_in = np.column_stack([latents, norms.reshape(-1,1), z_arr.reshape(-1,1)]).astype(np.float32)
    X_in = np.nan_to_num(X_in, nan=0.0)
    Y_tgt = np.nan_to_num(wise["mag"], nan=0.0).astype(np.float32)
    Y_err = wise["mag_err"].astype(np.float32)
    det   = wise["detected"]
    
    perm = np.random.permutation(N)
    n_te = int(N * 0.20)
    n_tr = N - n_te
    itr, ite = perm[:n_tr], perm[n_tr:]
    
    # ── Train MLP ────────────────────────────────────────────────────
    print(f"[4/5] Training MLP ({n_tr:,} train, {n_te:,} test) ...")
    mlp = WISEPredictor(8, 20, 2, 4).to(dev)
    opt = optim.Adam(mlp.parameters(), lr=1e-3)
    ds = TensorDataset(
        torch.tensor(X_in[itr], dtype=torch.float32),
        torch.tensor(Y_tgt[itr], dtype=torch.float32),
        torch.tensor(Y_err[itr], dtype=torch.float32),
    )
    loader = DataLoader(ds, batch_size=mlp_batch, shuffle=True)
    
    mlp.train()
    t0 = time.time()
    for ep in range(n_epochs):
        for xb, yb, eb in loader:
            xb, yb, eb = xb.to(dev), yb.to(dev), eb.to(dev)
            opt.zero_grad()
            gauss_nll(mlp(xb), yb, eb).backward()
            opt.step()
        if (ep+1) % 100 == 0:
            print(f"  epoch {ep+1}/{n_epochs}")
    print(f"  Done in {time.time()-t0:.1f}s")
    
    # ── Evaluate ─────────────────────────────────────────────────────
    print("[5/5] Evaluating ...")
    mlp.eval()
    with torch.no_grad():
        yp = mlp(torch.tensor(X_in[ite], dtype=torch.float32).to(dev)).cpu().numpy()
    
    # Save everything
    torch.save(mlp.state_dict(), out / "mlp_fullscale.pt")
    np.savez_compressed(out / "results.npz",
        y_pred=yp, y_true=Y_tgt[ite], y_err=Y_err[ite],
        detected=det[ite], latents=latents, norm_consts=norms)
    
    # Print summary
    for j, b in enumerate(["W1","W2","W3","W4"]):
        d = det[ite, j]
        if d.sum() > 0:
            chi2n = np.mean(((Y_tgt[ite,j][d] - yp[d,j]) / Y_err[ite,j][d])**2)
            print(f"  {b}: χ²_N = {chi2n:.3f}  (N_det={d.sum():,})")
    
    print(f"\nResults saved to {out}/")
    return out


# ── To actually run at full scale, uncomment: ────────────────────────
# run_full_scale(
#     sdss_dir="sdss_output",
#     wise_dir="wise_output",
#     out_dir="ppplot_output_fullscale",
#     device="cuda:0",          # or "cuda:1" for second H200
#     spender_batch=10_000,     # tune for your GPU VRAM
#     mlp_batch=4096,
# )

print("Full-scale function defined. Uncomment the call above to run.")
print(f"Expected runtime on your server: ~12 minutes for 510k galaxies.")
````

## Cell 17 — Parallel SDSS spectrum preprocessing at scale

This is the other bottleneck at 510k: downloading & preprocessing spectra.
Here's the fully-implemented parallel approach.

````python
def parallel_preprocess_sdss(
    meta_csv,
    out_dir="sdss_output_fullscale",
    n_jobs=64,
    chunk_size=500,
    wave_min=3800.0, wave_max=9200.0, n_pixels=3600,
    norm_min=5300.0, norm_max=5850.0,
):
    """
    Parallel SDSS spectrum download + preprocessing.
    
    For 510k spectra on a 768-core machine:
      - Download is I/O-bound (SDSS server), ~50-100 spec/min
      - Preprocessing is CPU-bound, ~10k spec/s with 64 workers
      - Total: ~8-16 hours for download, <1 min for preprocessing
    
    Better approach: bulk-download plate files from
      https://data.sdss.org/sas/dr17/sdss/spectro/redux/
    then process locally (see below).
    """
    from joblib import Parallel, delayed
    from scipy.interpolate import interp1d
    from pathlib import Path
    import pandas as pd
    
    out = Path(out_dir); out.mkdir(exist_ok=True)
    wave_grid = np.linspace(wave_min, wave_max, n_pixels)
    
    # Load metadata
    meta = pd.read_csv(meta_csv)
    print(f"Will process {len(meta):,} spectra using {n_jobs} workers")
    
    def process_one_spectrum(row):
        """Download + preprocess a single spectrum. Returns (flux, norm_const) or None."""
        try:
            from astroquery.sdss import SDSS
            sp = SDSS.get_spectra(
                plate=int(row["plate"]), mjd=int(row["mjd"]),
                fiberID=int(row["fiberid"])
            )
            if not sp: return None
            data = sp[0][1].data
            z = float(row["redshift"])
            wave_obs = 10.0 ** data["loglam"]
            flux = data["flux"].astype(np.float64)
            wave_rest = wave_obs / (1.0 + z)
            
            # Interpolate to grid
            valid = np.isfinite(flux) & (data["ivar"] > 0)
            if valid.sum() < 100: return None
            f_interp = interp1d(wave_rest[valid], flux[valid],
                                kind="linear", bounds_error=False, fill_value=0.0)
            flux_grid = f_interp(wave_grid)
            
            # Normalise
            nmask = (wave_grid >= norm_min) & (wave_grid <= norm_max)
            norm_vals = flux_grid[nmask]
            norm_vals = norm_vals[norm_vals > 0]
            if len(norm_vals) == 0: return None
            norm_const = float(np.median(norm_vals))
            if norm_const <= 0: return None
            flux_norm = flux_grid / norm_const
            
            return flux_norm.astype(np.float32), np.float32(norm_const)
        except Exception:
            return None
    
    # Process in parallel
    results = Parallel(n_jobs=n_jobs, verbose=10)(
        delayed(process_one_spectrum)(row) for _, row in meta.iterrows()
    )
    
    # Collect results
    good_idx = [i for i, r in enumerate(results) if r is not None]
    spectra = np.array([results[i][0] for i in good_idx], dtype=np.float32)
    norms   = np.array([results[i][1] for i in good_idx], dtype=np.float32)
    
    np.save(out / "spectra_matrix.npy", spectra)
    np.save(out / "norm_consts.npy", norms)
    np.save(out / "wave_grid.npy", wave_grid)
    
    print(f"Saved {len(good_idx):,}/{len(meta):,} spectra to {out}/")
    return out


print("""
FULL-SCALE SDSS APPROACH:

Option A — Bulk plate files (FASTEST, recommended):
  1. wget -r https://data.sdss.org/sas/dr17/sdss/spectro/redux/26/spectra/
     (or use rsync for selected plates)
  2. Each plate dir has a spPlate-PPPP-MMMMM.fits with 640 spectra
  3. Process locally with astropy.io.fits — no network bottleneck
  
Option B — Individual spectrum download (slow but simple):
  parallel_preprocess_sdss(
      "sdss_metadata.csv",  # from CasJobs
      n_jobs=64,            # tune for your server
  )

Either way saves spectra_matrix.npy + norm_consts.npy.
Then run_full_scale() picks them up.
""")
````
