# SDSS Spectra: Fetching, Preprocessing & Clustering
### Starting point for reproducing Jespersen et al. (2025) — *"The optical and infrared are connected"* ([arXiv:2503.03816](https://arxiv.org/abs/2503.03816))

**What this notebook does:**
1. Queries the SDSS SkyServer for a sample of Main Galaxy Sample spectra
2. Downloads and preprocesses them onto a common rest-frame wavelength grid
3. Builds a clean `numpy` array ready for clustering / dimensionality reduction
4. Runs a PCA + k-means clustering demo and visualises the results

**The paper pipeline:** SDSS spectra → `spender` autoencoder (6-D latents) → small MLP → predicted WISE IR photometry. This notebook reproduces the data ingestion and preprocessing, with PCA as a transparent stand-in for `spender`. The final cell explains how to swap in `spender` to match the paper exactly.

---
> **Scale knob:** Change `N_SPECTRA` in Cell 2 to go from a quick 50-spectrum test all the way up to the full ~510,000 object dataset.

## Cell 0 — Imports

````python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from pathlib import Path
from scipy.interpolate import interp1d

# Astronomy
from astropy import units as u
from astropy.table import Table
from astroquery.sdss import SDSS
from specutils import Spectrum1D

# ML
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.manifold import TSNE

%matplotlib inline
plt.rcParams.update({"figure.dpi": 120, "font.size": 11})

print("All imports OK ✓")
````

## Cell 1 — Configuration
Edit `N_SPECTRA` to control how many galaxies to download. Start with a small number (e.g. 50–200) to verify everything works, then scale up.

````python
# ── How many spectra to download ────────────────────────────────────────────
# Start small (50–200) for a quick test; scale to ~500k for the full paper.
N_SPECTRA = 200

# ── Rest-frame wavelength grid (Angstroms) ───────────────────────────────────
# Covers the full SDSS optical range at ~1.5 Å / pixel (R ~ 2000)
WAVE_MIN  = 3800.0
WAVE_MAX  = 9200.0
N_PIXELS  = 3600

# ── Normalisation window ─────────────────────────────────────────────────────
# Paper normalises each spectrum by the median flux in this V-band window
NORM_WAVE_MIN = 5300.0
NORM_WAVE_MAX = 5850.0

# ── Redshift limit matching the paper ────────────────────────────────────────
Z_MAX = 0.5

# ── SDSS sky-emission lines to mask (observer-frame, Angstroms) ──────────────
SKYLINES_OBS    = np.array([
    5577.3, 5889.9, 6157.0, 6300.3, 6363.8,
    6834.3, 6864.0, 7340.9, 7523.7, 7715.0,
    8344.6, 8399.0, 8430.0, 8791.2, 8827.1, 9375.0,
])
SKYLINE_WIDTH_AA = 10.0   # half-width in Å

# ── Output directory ─────────────────────────────────────────────────────────
OUT_DIR = Path("sdss_output")
OUT_DIR.mkdir(exist_ok=True)

# ── Build rest-frame grid ────────────────────────────────────────────────────
WAVE_GRID = np.linspace(WAVE_MIN, WAVE_MAX, N_PIXELS)

print(f"Config ready. Will fetch {N_SPECTRA} spectra, "
      f"grid: {WAVE_MIN:.0f}–{WAVE_MAX:.0f} Å  ({N_PIXELS} pixels)")
````

## Cell 2 — Query SDSS SkyServer for galaxy metadata

We query against `SpecObj` joined with `PhotoObj`, applying the same selection as the paper:
- `class = GALAXY` (removes stars and QSOs)
- `z` between 0.01 and 0.5
- `zWarning = 0` (clean, unflagged redshift)
- `petroMag_r < 17.8` (Main Galaxy Sample bright limit)

````python
def query_sdss_galaxies(n=N_SPECTRA, z_max=Z_MAX):
    """
    SQL query against the SDSS SkyServer.
    Returns an astropy Table with plate/mjd/fiberid + basic photometry.
    For the full ~510k sample, remove the TOP {n} clause and run via CasJobs.
    """
    sql = f"""
    SELECT TOP {n}
        s.specobjid,
        s.plate,
        s.mjd,
        s.fiberid,
        s.z           AS redshift,
        s.zwarning,
        s.class       AS specclass,
        s.subclass    AS specsubclass,
        p.petroMag_r,
        p.petroMag_g,
        p.petroMag_i
    FROM
        SpecObj AS s
    JOIN
        PhotoObj AS p ON s.bestobjid = p.objid
    WHERE
        s.class    = 'GALAXY'
        AND s.z    BETWEEN 0.01 AND {z_max}
        AND s.zwarning = 0
        AND p.petroMag_r BETWEEN 14.0 AND 17.8
        AND p.petroMag_r != -9999
    ORDER BY
        s.plate, s.mjd, s.fiberid
    """
    result = SDSS.query_sql(sql, timeout=120)
    if result is None or len(result) == 0:
        raise RuntimeError("Query returned no results — check network / SDSS service.")
    return result

print("Querying SDSS SkyServer …")
meta_table = query_sdss_galaxies()
print(f"Retrieved {len(meta_table)} objects")
meta_table[:5]
````

## Cell 3 — Download spectra

Each SDSS spectrum FITS file contains:
- `loglam` — log₁₀(wavelength in Å), log-spaced at ~10⁻⁴ dex/pixel
- `flux` — flux density in units of 10⁻¹⁷ erg/s/cm²/Å
- `ivar` — inverse variance (zero = bad pixel)
- `andmask` — bitmask (non-zero = some data quality issue)

````python
def download_spectra(meta):
    spectra, failed = [], 0
    for i, row in enumerate(meta):
        try:
            sp = SDSS.get_spectra(
                plate=int(row["plate"]),
                mjd=int(row["mjd"]),
                fiberID=int(row["fiberid"]),
            )
            if not sp:
                failed += 1
                continue
            data = sp[0][1].data
            spectra.append({
                "specobjid"   : row["specobjid"],
                "redshift"    : float(row["redshift"]),
                "specsubclass": str(row["specsubclass"]).strip(),
                "petroMag_r"  : float(row["petroMag_r"]),
                "wavelength"  : (10.0 ** data["loglam"]).astype(np.float32),
                "flux"        : data["flux"].astype(np.float32),
                "ivar"        : data["ivar"].astype(np.float32),
                "andmask"     : data["andmask"].astype(np.int32),
            })
        except Exception as e:
            failed += 1
            if failed <= 3:
                print(f"  Warning row {i}: {e}")
        if (i + 1) % 50 == 0:
            print(f"  {i+1}/{len(meta)} …")

    print(f"Downloaded {len(spectra)} spectra  ({failed} failed)")
    return spectra

spectra_raw = download_spectra(meta_table)
````

## Cell 4 — Preprocessing: deredshift → common grid → normalise → mask sky lines

Steps applied to every spectrum:

| Step | Operation | Why |
|------|-----------|-----|
| 1 | `λ_rest = λ_obs / (1 + z)` | Put all spectra in the same rest frame |
| 2 | Mask `andmask ≠ 0`, `ivar ≤ 0`, sky lines | Remove unreliable pixels |
| 3 | Linear interpolation onto WAVE_GRID | Common pixel basis for matrix operations |
| 4 | Divide by median flux in 5300–5850 Å | Paper's normalisation constant |
| 5 | Clip at ±10 × nMAD | Remove cosmic rays / reduction artefacts |

````python
def preprocess_spectrum(spec, wave_grid=WAVE_GRID, fill_value=0.0):
    z          = spec["redshift"]
    wave_obs   = spec["wavelength"].astype(np.float64)
    flux       = spec["flux"].astype(np.float64)
    ivar       = spec["ivar"].astype(np.float64)
    andmask    = spec["andmask"]

    # 1. Deredshift
    wave_rest = wave_obs / (1.0 + z)

    # 2. Bad-pixel mask
    bad = (andmask != 0) | (ivar <= 0) | (~np.isfinite(flux))
    for sky_lam in SKYLINES_OBS:
        bad |= (np.abs(wave_rest - sky_lam / (1.0 + z)) < SKYLINE_WIDTH_AA)
    flux_clean       = flux.copy()
    flux_clean[bad]  = np.nan

    # 3. Interpolate onto common grid
    valid = ~np.isnan(flux_clean)
    if valid.sum() < 100:
        return None
    try:
        interp_fn = interp1d(
            wave_rest[valid], flux_clean[valid],
            kind="linear", bounds_error=False, fill_value=fill_value,
        )
        flux_grid = interp_fn(wave_grid)
    except Exception:
        return None

    # 4. Normalise
    norm_mask = (wave_grid >= NORM_WAVE_MIN) & (wave_grid <= NORM_WAVE_MAX)
    norm_vals = flux_grid[norm_mask]
    norm_vals = norm_vals[norm_vals > 0]
    if len(norm_vals) == 0:
        return None
    norm_const = np.median(norm_vals)
    if not np.isfinite(norm_const) or norm_const <= 0:
        return None
    flux_norm = flux_grid / norm_const

    # 5. Clip outliers (robust: median ± 10 nMAD)
    fm = np.isfinite(flux_norm) & (flux_norm != fill_value)
    if fm.sum() == 0:
        return None
    med  = np.median(flux_norm[fm])
    nmad = 1.4826 * np.median(np.abs(flux_norm[fm] - med))
    flux_norm = np.clip(flux_norm, med - 10 * nmad, med + 10 * nmad)

    return flux_norm.astype(np.float32)


# Apply to all downloaded spectra
rows, meta_clean = [], []
for spec in spectra_raw:
    v = preprocess_spectrum(spec)
    if v is not None:
        rows.append(v)
        meta_clean.append({
            "specobjid"   : spec["specobjid"],
            "redshift"    : spec["redshift"],
            "specsubclass": spec["specsubclass"],
            "petroMag_r"  : spec["petroMag_r"],
        })

X         = np.vstack(rows)      # ← THE DATA MATRIX
redshifts = np.array([m["redshift"] for m in meta_clean])
subclass  = np.array([m["specsubclass"] for m in meta_clean])

print(f"\nData matrix X: {X.shape}   (spectra × pixels)")
print(f"dtype: {X.dtype}    min: {X.min():.3f}    max: {X.max():.3f}")
print(f"Pass rate: {len(rows)}/{len(spectra_raw)} ({len(rows)/len(spectra_raw)*100:.1f}%)")
````

### Quick-look: same spectra after preprocessing

````python
fig, axes = plt.subplots(3, 1, figsize=(12, 7), sharex=True)

# Key spectral lines to mark
lines = {"[OII]": 3727, "Hβ": 4861, "[OIII]": 5007,
         "MgI": 5175, "NaI": 5893, "Hα": 6563, "[SII]": 6717}

for ax, flux_vec, m in zip(axes, X[:3], meta_clean[:3]):
    ax.plot(WAVE_GRID, flux_vec, lw=0.7, color="#2c7bb6")
    for name, lam in lines.items():
        if WAVE_MIN < lam < WAVE_MAX:
            ax.axvline(lam, ls=":", lw=0.8, color="firebrick", alpha=0.6)
            ax.text(lam + 15, ax.get_ylim()[1] * 0.8, name,
                    fontsize=7, color="firebrick", rotation=90)
    ax.set_ylabel("Norm. flux")
    ax.set_title(f"z = {m['redshift']:.4f}   subclass = {m['specsubclass'] or 'None'}")

axes[-1].set_xlabel("Rest-frame wavelength (Å)")
plt.suptitle("Preprocessed rest-frame spectra (normalised, sky-masked)",
             fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
````

## Cell 6 — PCA: spectral compression

The paper uses `spender`'s **6-D autoencoder latent space** as the compressed representation fed into the MLP. PCA is used here as a transparent, dependency-free alternative — it makes the spectral variance structure easy to inspect before committing to `spender`.

The scree plot below shows how quickly variance accumulates in the leading components.

````python
# StandardScaler: zero mean, unit variance per pixel
scaler  = StandardScaler()
X_scaled = scaler.fit_transform(np.nan_to_num(X, nan=0.0))

# PCA
N_PC = min(50, X.shape[0] - 1)
pca   = PCA(n_components=N_PC, random_state=42)
X_pca = pca.fit_transform(X_scaled)

explained_cumulative = pca.explained_variance_ratio_.cumsum()
print(f"First  6 PCs explain {explained_cumulative[5]*100:.1f}% of variance  "
      f"(spender latent dim)")
print(f"First 50 PCs explain {explained_cumulative[-1]*100:.1f}% of variance")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Scree: individual explained variance
ax = axes[0]
ax.bar(range(1, N_PC + 1), pca.explained_variance_ratio_ * 100,
       color="steelblue", alpha=0.8)
ax.set_xlabel("Principal component")
ax.set_ylabel("Variance explained (%)")
ax.set_title("Scree plot")
ax.set_xlim(0, min(20, N_PC) + 1)

# Cumulative
ax = axes[1]
ax.plot(range(1, N_PC + 1), explained_cumulative * 100,
        "o-", color="darkorange", ms=4, lw=1.5)
ax.axvline(6, ls="--", color="gray", lw=1.2,
           label=f"spender dim = 6 ({explained_cumulative[5]*100:.0f}%)")
ax.set_xlabel("Number of PCs")
ax.set_ylabel("Cumulative variance explained (%)")
ax.set_title("Cumulative variance")
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

### Visualise the leading eigenvectors (eigenspectra)

````python
fig, axes = plt.subplots(3, 2, figsize=(14, 8), sharex=True)
for i, ax in enumerate(axes.flat):
    ax.plot(WAVE_GRID, pca.components_[i], lw=0.9, color=f"C{i}")
    ax.axhline(0, ls="--", lw=0.6, color="gray", alpha=0.5)
    ax.set_title(f"PC {i+1}  ({pca.explained_variance_ratio_[i]*100:.1f}% variance)")
    ax.set_ylabel("Loading")
    for name, lam in lines.items():
        if WAVE_MIN < lam < WAVE_MAX:
            ax.axvline(lam, ls=":", lw=0.7, color="gray", alpha=0.5)
for ax in axes[-1]:
    ax.set_xlabel("Rest-frame wavelength (Å)")
plt.suptitle("Leading PCA eigenspectra", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
````

## Cell 7 — k-means clustering in PCA space

We cluster in the 6-D PCA sub-space to match the dimensionality of the `spender` latents used in the paper. Change `N_CLUSTERS` to explore different numbers of galaxy groups.

````python
N_CLUSTERS = 6   # matches the number of spender latent dimensions

km = KMeans(n_clusters=N_CLUSTERS, random_state=42, n_init=10)
labels = km.fit_predict(X_pca[:, :6])   # use first 6 PCs

unique, counts = np.unique(labels, return_counts=True)
print("Cluster sizes:")
for k, c in zip(unique, counts):
    subclasses_in_k = subclass[labels == k]
    dominant = "—"
    if len(subclasses_in_k) > 0:
        vals, cnts = np.unique(subclasses_in_k, return_counts=True)
        dominant = vals[cnts.argmax()]
    print(f"  Cluster {k}: {c:4d} galaxies  (dominant subclass: {dominant or 'None'!r})")
````

### Cluster mean spectra

````python
colors = cm.tab10(np.linspace(0, 1, N_CLUSTERS))

fig, axes = plt.subplots(N_CLUSTERS, 1, figsize=(13, 3 * N_CLUSTERS),
                          sharex=True)
for k, (ax, col) in enumerate(zip(axes, colors)):
    mask = labels == k
    centroid = X[mask].mean(axis=0)
    ax.plot(WAVE_GRID, centroid, color=col, lw=1.2)
    ax.fill_between(WAVE_GRID,
                    centroid - X[mask].std(axis=0),
                    centroid + X[mask].std(axis=0),
                    color=col, alpha=0.15)
    for name, lam in lines.items():
        if WAVE_MIN < lam < WAVE_MAX:
            ax.axvline(lam, ls=":", lw=0.7, color="gray", alpha=0.6)
    ax.set_ylabel("Norm. flux")
    ax.set_title(f"Cluster {k}  (N = {mask.sum()})  ±1σ envelope",
                 color=col, fontweight="bold")
    ax.set_xlim(WAVE_MIN, WAVE_MAX)
axes[-1].set_xlabel("Rest-frame wavelength (Å)")
plt.suptitle("Cluster mean spectra ± 1σ", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
````

### PC1 vs PC2 scatter, coloured by cluster

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# PC1 vs PC2
ax = axes[0]
for k, col in enumerate(colors):
    mask = labels == k
    ax.scatter(X_pca[mask, 0], X_pca[mask, 1],
               c=[col], s=12, alpha=0.65, label=f"Cluster {k}")
ax.set_xlabel("PC 1")
ax.set_ylabel("PC 2")
ax.set_title("PC1 vs PC2 — coloured by k-means cluster")
ax.legend(fontsize=9, markerscale=2)

# PC1 vs PC2 coloured by redshift
ax = axes[1]
sc = ax.scatter(X_pca[:, 0], X_pca[:, 1],
                c=redshifts, cmap="plasma", s=12, alpha=0.7)
plt.colorbar(sc, ax=ax, label="Redshift z")
ax.set_xlabel("PC 1")
ax.set_ylabel("PC 2")
ax.set_title("PC1 vs PC2 — coloured by redshift")

plt.suptitle("PCA projection of SDSS spectra", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
````

### t-SNE projection (optional — slow for large N)

````python
# t-SNE is slow for N > ~5000; subsample if needed
MAX_TSNE = min(X.shape[0], 500)
idx_tsne = np.random.choice(X.shape[0], MAX_TSNE, replace=False)

tsne  = TSNE(n_components=2, perplexity=30, random_state=42, n_iter=1000)
X_tsne = tsne.fit_transform(X_pca[idx_tsne, :20])   # feed 20 PCs to t-SNE

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for ax, c_vals, title, cmap in [
    (axes[0], labels[idx_tsne], "k-means cluster", "tab10"),
    (axes[1], redshifts[idx_tsne], "Redshift", "plasma"),
]:
    sc = ax.scatter(X_tsne[:, 0], X_tsne[:, 1],
                    c=c_vals, cmap=cmap, s=10, alpha=0.7)
    plt.colorbar(sc, ax=ax, label=title)
    ax.set_xlabel("t-SNE 1")
    ax.set_ylabel("t-SNE 2")
    ax.set_title(f"t-SNE — coloured by {title}")

plt.suptitle(f"t-SNE projection (N = {MAX_TSNE} subsample)",
             fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
````

## Cell 10 — Next steps: swapping in `spender` (the paper's actual encoder)

The paper compresses each SDSS spectrum to a **6-D latent vector** using [spender](https://github.com/pmelchior/spender), a neural spectral autoencoder trained on the full SDSS Main Galaxy Sample.  The `spender` latents + redshift + normalisation constant = 8 inputs to the tiny MLP that predicts WISE photometry.

```python
# Install
# conda install pytorch -c pytorch   (or pip install torch)
# pip install spender

import torch
import spender

# Load pretrained model (~50 MB download, cached automatically)
model, data_class = spender.load("SDSS-main")
model.eval()

# X has shape (N, 3600) — the matrix we already built
flux_tensor = torch.tensor(X, dtype=torch.float32)
wave_tensor = torch.tensor(WAVE_GRID, dtype=torch.float32)

with torch.no_grad():
    latents, norm_consts = model.encode(flux_tensor, wave_tensor)
# latents.shape == (N, 6)  ← drop-in replacement for X_pca[:, :6]

# Reconstruction (to check quality)
with torch.no_grad():
    X_reconstructed = model.decode(latents, wave_tensor).numpy()
```

### Scaling to the full ~510,000 object dataset

```python
# 1. Remove "TOP {n}" from the SQL query for the full MGS
# 2. Use SDSS CasJobs (https://skyserver.sdss.org/casjobs/) for large queries
# 3. Bulk-download plate files:
#    https://data.sdss.org/sas/dr17/sdss/spectro/redux/
# 4. WISE cross-match (from SDSS SkyServer):
#    SELECT * FROM wise_allsky    -- WISE photometry
#    SELECT * FROM wise_xmatch    -- 2" astrometric cross-match
# 5. After encoding with spender, train the MLP:
#    inputs  = [latents (6), norm_const (1), redshift (1)]  → 8 features
#    targets = WISE W1, W2, W3, W4 Vega magnitudes
#    loss    = Gaussian NLL  (Eq. 1 of the paper)
# Demo notebook from the paper authors:
#    https://github.com/astrockragh/IR_optical_demo
```
